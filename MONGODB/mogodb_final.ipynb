```python
##Insert
std_id=input("Enter ur id : ")
name =input("Enter ur name : ")
marks=int(input("Enter ur marks : "))
dept=input("Enter ur dept : ")

coll.insert_one({"id":std_id,"name":name,"marks":marks,"department":dept})
print("Data is inserted successfully !!")
```

```python
### read
for data in coll.find():
  print(data)
```

```python
## sorting data based on marks ASC
for data in coll.find().sort("marks",1):
  print(data)
```

```python
## sorting data based on marks DESC
for data in coll.find().sort("marks",-1):
  print(data)
```

```python
## limit clause
for data in coll.find().sort("marks",-1).limit(2):
  print(data)
```

```python
##highest marks in database
for data in coll.find().sort("marks",-1).limit(1):
  print(data)
```

```python
##second highest marks in database
for data in coll.find().sort("marks",-1).limit(1).skip(1):
  print(data)
```

```python
##Third highest marks in database
for data in coll.find().sort("marks",-1).limit(1).skip(2):
  print(data)
```

```python
## lowest marks in database
for data in coll.find().sort("marks",1).limit(1):
  print(data)
```
