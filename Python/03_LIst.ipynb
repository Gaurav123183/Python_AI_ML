Packing And Unpacking

```python
lst=[10,20,30] ## -> lst ke andar sabhi ko pack ->packing
a,b,c=lst ## unpack a=10 , b =20 , c=30
print(a)
print(b)
print(c)
```

```python
lst=[10,20,30]
a,b=lst
print(a)
print(b)
```

```python
lst=[10,20,30]
*a,b=lst
print(a)
print(b)
```

```python
lst=["a","b","c"]
x,y=lst
print(x)
print(y)
```

```python
lst=["a","b","c"]
x,y,z=lst
print(x)
print(y)
print(z)
```

```python
##swap -> without function -> for loop
##a=10 b=20   -> a=20 , b=10 lst=[10,20,30,40]

lst=[10,20,30,40] ### lst ->[40,20,30,10]
   # 0 ,1 , 2, 3
   #.-4,-3,-2,-1
print(f"Before swaping lst[0] -> {lst[0]}")

lst[0],lst[-1]=lst[-1],lst[0]
print(f"After swaping lst[0] -> {lst[0]}")

print(lst[-1])
```
