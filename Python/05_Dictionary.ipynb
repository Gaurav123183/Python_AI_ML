Dictonary comprehension

```python
sqr={
    x: x*x for x in range(11)




}
```

```python
print(sqr)
```

```python
##even odd
even={
    x : "even" if x%2==0 else "odd" for x in range(1,1000)
}
```

```python
print(even)
```

```python
dict={
    "xyz":25,
    "abc":35,
    "mno":45
}

result={
    name:"Pass" if marks>=30 else "Fail"
    for name , marks in dict.items()

}
```

```python
print(result)
```
