Arithematic operators

```python
a=10
b=20

print(a+b)
print(a-b)
print(a*b)
print(a/b)
#floor-> //
print(a//b) #-> use to print queotient
print(a%b) # -> use to print remainder
print(a**b)
print(2**3)
print(2**4)
```

**comparision operators**

```python
a=10
b=20

print(a==b)
print(a>b)
print(a<b)
print(a!=b)
print(a>=b)
print(a<=b)
```

Logical operator

```python
#and
#or
#not
a=10
b=20

print(a<15 and b<25)
print(a>5 or b>30)
print(not(a<b))
```

Bitwise operator

```python
a=5
b=3
print(a&b)
print(a|b)
print(a^b)
```
