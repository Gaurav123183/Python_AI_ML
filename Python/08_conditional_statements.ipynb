```python
password=input("Enter Your password : ")

if len(password)<8:
  print("Invalid")
else :
   has_digit=False
   has_lower=False
   has_upper=False

   for ch in password:
    if ch.islower():
      has_lower=True
    elif  ch.isdigit():
      has_digit=True
    elif ch.isupper():
      has_upper=True
   if has_lower and has_upper and has_digit:
      print("Your password is valid")
   else:
      print("Your password is invalid")

```
