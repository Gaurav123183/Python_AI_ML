```python
num=1234
rev=0
while num>0:
  digit=num%10
  rev=rev*10+digit ##rev=0  digit=4  -> 0*10+4=4
  num//=10 ## 1234 // 10 -> 123 // 10 -> 12 //10 -> 1 // 10 -> 0.1 stopping condition
print(rev)
```

Factorial

```python
num=8
fact=1
for i in range(1,num+1):
  fact*=i
print(fact)

#fact =1
# i=1 -> fact=1
# i=2 -> fact=2
# i=3 -> fact=6
# i=4 -> fact=24
# i=5 -> fact=120
```
