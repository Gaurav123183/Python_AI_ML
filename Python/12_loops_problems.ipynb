```python
num=1234
original=num ## original=121
rev=0
while num>0:
  digit=num%10
  rev =rev *10 + digit
  num//=10
if original==rev:
  print("Palindrome")
else:
  print("Not Palindrome")

```

Largest Digit

```python
num=123598
largest=0
while num>0:
  digit=num%10
  if digit > largest:
    largest=digit
  num//=10
print(largest)
```

Smallest digit

```python
num=123598
smallest=num #123598 -> smallest= 8 -> 9 -> smallest=5  3  ->smallest=1
while num>0:
  digit=num%10
  if digit < smallest:
    smallest=digit
  num//=10
print(smallest)
```
