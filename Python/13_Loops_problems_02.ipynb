```python
lst=[1,2,3,4,6,8]
    #0,1,2,3,4,5
    #i.        j
target=10
i=0
j=len(lst)-1
while i<j:
  sum=lst[i]+lst[j]
      ## lst[0]+lst[5]->1 + 8
  if target==sum:
    print(lst[i],lst[j])
    break
  elif sum >target:
    j-=1
  else :
    i+=1
```

Nested loop

```python
for i in range(11): #0 -> 3  start 0 1 2
    for j in range(11):
      print("*", end=" ")
    print()
```

```python
for i in range(4): #0 -> 3  start 0 1 2
    for j in range(i):
      print("*",end=" ")
    print()
```

```python
for i in range(5,0,-1):
    for j in range(i):
      print("*",end=" ")
    print()
```
