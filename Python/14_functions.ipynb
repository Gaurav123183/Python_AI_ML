one parameter

Two parameters

```python
def add(a,b):
  print(a+b)
add(10,20)
```

```python
def add(a,b):
  return a+b
result=add(100,200)
print(result)
```

```python
def calulate(a,b):
  return a*b,a+b,a-b,(a+b)+5 ##-> packing
x , y , z ,m = calulate(10,20) ##-> unpacking
print(x)
print(y)
print(z)
print(m)
```

```python
# num % 2 ==0 -> even -> odd
def check(num):
  if num % 2==0 :
    return "Even "
  else:
    return "odd"
check(101)
```

Count vowels

```python
def count_vowels(text):
     count=0
     for char  in text:
       if char in "aeiou":
        count+=1
     return count
ans=count_vowels("python")
print(ans)
```

Frequency count -> PYQ

```python
def freq(text):
  dict={}

  for char in text:
    if char in dict:
      dict[char]+=1
    else :
      dict[char]=1
  return dict
print(freq("python programming"))
```
