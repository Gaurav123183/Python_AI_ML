SQUARE PATTERN

```python
for i in  range(1,6): ## start -> end or end -> start
  for j in range(1,6):
   print("*",end=" ")
  print()
```

Rectangle Pattern

```python
for i in range(1,4):
  for j in range(1,6):
    print("*",end=" ")
  print()
```

```python
for i in range(1,6):
  for j in range(1,4):
    print("*",end=" ")
  print()
```

```python
for i in range(1,1):
  print("*")
```

Right angled triangle

```python
for i in range(1,6): # i=2, i=3 , i=4
  for j in range(1,i+1): ## range(1,2)-> 1,, (1,3)-> 2. ,(1,4)-> 3 ,(1,5)-> 4
   print("*",end=" ")
  print()
```

```python
for i in range(5,0,-1):  ## i=5 ,i=4, i=3 i=2 i=1 i=0-> loop stop
  for j in range(1,i+1): ## range(1,6)-> 5
                         ## range(1,5)-> 4
                         ## range(1,4)-> 3
                         ## range(1,3)-> 2
                         ## range(1,2)-> 1

   print("*",end=" ")
  print()
```

Right aligned triangle

```python
for i in range(1,5): ## 4
  ## spaces
  for j in range(1,4-i+1):
    print(" ",end=" ")
  ##star
  for j in range(1,i+1):
    print("*",end=" ")
  print()
```

```python
n =int(input("Enter number of rows u want to print : "))
for i in range(1,n+1): ## 4
  ## spaces
  for j in range(1,n+1-i+1):
    print(" ",end=" ")
  ##star
  for j in range(1,i+1):
    print("*",end=" ")
  print()
```

Hollow Square

```python
for i in range(1,5):
  for j in range(1,5):
    if i==1 or i==4 or j==1 or j==4:
     print("*",end=" ")
    else:
      print(" ",end=" ")
  print()
```

Hollow right angled triangle


```python
for i in range(1,5):
  for j in range(1,i+1):
    if j==1 or i==4 or j==i:
      print("*",end=" ")
    else:
      print(" ",end=" ")
  print()
```
