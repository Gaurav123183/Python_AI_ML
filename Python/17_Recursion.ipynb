```python
def fun(n):
  for i in range(1,n+1):
    print(i)
fun(5)
```

Print 1->N by using recursion

```python
def fun(n):

  ##base case -> stop kaha hona he
  if n==0:
    return

## recursive case -> function repeated call karta he n -> n-1
  fun(n-1)

  print(n)


fun(5)

```

Sum of N Numbers


```python
def sum(n):
  if n==0:
    return 0

  return n+sum(n-1)

ans=sum(5)
print(ans)
```

Multiplication of N number

```python
def multi(n):
  if n==0:
    return 1

  return n * multi(n-1)

ans=multi(5)
print(ans)
```

Factorial of n Numbers


```python
def fact(n):
  if n==0:
    return 1

  return n * fact(n-1)

n=int(input("enter the number : "))
print(f"Factorial of {n} is : ",fact(n))
```
