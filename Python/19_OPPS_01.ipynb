```python
class student:

  def __init__(self,name,age,rollno):
   self.name=name
   self.age=age
   self.rollno=rollno

  def display(self):
    print("Name is ",self.name,"Age is ",self.age,"Rollno is ",self.rollno)

s1=student("Gaurav",21,23)

s1.display()


```
