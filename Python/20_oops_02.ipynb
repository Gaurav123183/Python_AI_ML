```python
class employee :
  ##constructor
  def __init__(self,name,age,salary,company):
    self.name=name
    self.age=age
    self.salary=salary
    self.company=company

##object
e1=employee("Gaurav",21,20000,"TCS")
print(e1.name)
print(e1.company)
```

```python
class student:
  def __init__(self,name,age,rollno):
    self.name=name
    self.age=age
    self.rollno=rollno

  def display(self):
    print("Name is ",self.name," Age is  : ",self.age," Rollno is : ",self.rollno)

#object create
s=student("Xyz",23,90)

print(s.name)
print(s.age)

##class -> display

s.display()
```

```python
class bank:
  def __init__(self , name ,balance):
    self.name=name
    self.balance=balance

  def deposit(self,amount):
     self.balance+=amount

  def display(self):
    print("Name of customer is : ",self.name," Balance is : ",self.balance)

#object

b1=bank("Gaurav",2000)
print("Balance before deposit")
b1.display()

b1.deposit(10000)
print("Balance After deposit")

b1.display()

b1.deposit(2000)

b1.display()

```

Area of Rectangle

```python
class rectangle:
  def __init__(self ,length,width):
    self.length=length
    self.width=width
    self.area=length * width

#obje
r1=rectangle(100,99)
print("Area of rectangle having length",r1.length," and width ",r1.width," is",r1.area)
```

Area of Circle

```python
class circle:
  def __init__(self,radius):
    self.radius=radius
    self.area=3.14*radius*radius

c1=circle(5)
print("area of circle is : ",c1.area)
```

```python
class calculator:
  def __init__(self,a,b):
    self.a=a
    self.b=b

  def add(self):
    return self.a+self.b

  def sub(self):
    return self.a-self.b

  def multiply(self):
    return self.a*self.b

  def swap(self):
     self.a=self.a+self.b
     self.b=self.a-self.b
     self.a=self.a-self.b
     return self.a,self.b

cal=calculator(10,20)


cal.sub()

```

Destructor

```python

class student:

  def __init__(self,name):
    self.name=name
    print(self.name," is created")


  def __del__(self):
    print(self.name," is destroyed")

s1=student("Gaurav")


del s1

s2=student("Gaurav")
del s2

```
