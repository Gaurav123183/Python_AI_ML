Public member -> Everyone can access

```python
class student:
  def __init__(self,name):
    self.name=name

s1=student("Raju")
print(s1.name)
```

```python
class car:
  def __init__(self,car_name,color):
    self.car_name=car_name
    self.color=color

obj1=car("BMW","Black")
print(obj1.car_name)
print(obj1.color)
```

```python
class employee:
  def __init__(self,name,salary):
    self.name=name
    self.salary=salary

e1=employee("Raju",50000)
print(e1.name)
print(e1.salary)
```

Protected Class -> Everyone cannot access

```python
class bank:
  def __init__(self,name,balance):
    self.name=name        ##self.variable_name-> public
    self._balance=balance ## self._variable_name ->protected

b1=bank("abc",5000)
print(b1.name)
print(b1.balance)
```

```python
class bank:
  def __init__(self,name,balance):
    self.name=name        ## self.variable_name-> public
    self._balance=balance ## self._variable_name ->protected

b1=bank("abc",5000)
print(b1.name)
print(b1._balance)
```

```python
class student:
  def __init__(self , name ,backlog):
    self.name=name
    self._backlog=backlog

s1=student("Raju",3)
print(s1.name)
print(s1._backlog)
```

```python
class marks:
  def __init__(self,name,marks):
    self.name=name
    self._marks=marks

m1=marks("Raju",90)
print(m1.name)
print(m1._marks)
```

Private member -> Not one can able to access it easily

```python
class student:
  def __init__(self,name,marks):
    self.name=name ## public
    self.__marks=marks ## private

s1=student("Raju",50)
print(s1.name)

print(s1.__marks)

```

```python
class student:
  def __init__(self,name,marks):
    self.name=name
    self.__marks=marks


  def display(self):
    print(self.__marks)
s1=student("Raju",50)
print(s1.name)
s1.display()
```

```python
class std:
  def __init__(self):
    self.__marks=90


  def show(self):
    print(self.__marks)

s1=std()
s1.show()
```

```python
class bank:
    def __init__(self):
      self.__balance=4000

    def get_balance(self):
      return self.__balance

b1=bank()
print(b1.get_balance())
```

```python
class atm:
  def __init__(self):
    self.__pin=1234

  def check(self,password):
    return self.__pin==password

a1=atm()
print(a1.check(1234))
```
