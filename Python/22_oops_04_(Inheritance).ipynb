```python
class student:
   def study(self):
    print("student is studying ")
   def class_attend(self):
    print("Student is attending class")

class Enggstudent(student):
  def coding(self):
    print("Student is coding ")

e1=Enggstudent()
e1.study()
e1.class_attend()
e1.coding()
```

1 -> single inheritance
one class and one parent

```python
## A -> B
class A:
  def display(self):
    print("Me parent hu re ")

class B(A):
  def show(self):
    print("Me child hu re ")

b1=B()
b1.display()
b1.show()
```

```python
class Animal:
  def eat(self):
    print("Eating")

class Dog(Animal):
  def bark(self):
    print("Barking")
d1=Dog()
d1.eat()
d1.bark()
```

2 -> multilevel inheritance
A -> B -> C

```python
class A:
  def display(self):
    print("this is A")
class B(A):
  def show(self):
    print("this is B")
class C(B):
  def c_ka_show(self):
    print("this is C ")

c1=C()
c1.display()
c1.show()
c1.c_ka_show()
```

```python
class Animal:
  def eat(self):
    print("Eating")

class Dog(Animal):
  def bark(self):
    print("Barking")

class puppy(Dog):
  def play(self):
    print("Playing")
p1=puppy()
p1.eat()
p1.bark()
p1.play()
```

3 -> Multiple Inheritance
 ### A   B
 ###  \ /
   #   C

```python
class A:
  def display(self):
    print("This is A")
class B:
  def show(self):
    print("This is B")
class C(A,B):
  def show_c(self):
    print("THis is C")

c1=C()
c1.display()
c1.show()
c1.show_c()
```

```python
class father:
  def paisa(self):
    print("paisa")

class mother:
  def khana(self):
    print("khana")

class child(father,mother):
  def dono(self):
    print("Paisa and khana ")

c1=child()
c1.paisa()
c1.khana()
c1.dono()
```
