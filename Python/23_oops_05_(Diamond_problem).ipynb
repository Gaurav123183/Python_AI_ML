Hybrid Inheritance

```python
class A:
  def A_display(self):
    print("THis is A")
class B (A):
  def B_display(self):
    print("This is B")
class C(A):
  def C_display(self):
    print("THis is C")
class D(B,C):
  pass

d1=D()
d1.A_display()
d1.B_display()
d1.C_display()
```
