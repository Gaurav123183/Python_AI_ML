Dunder Function

Addition

```python
class Num:
  def __init__(self, val):
    self.val=val
  ##addition

  def __add__(self , obj):
    return self.val+obj.val
##object create
a=Num(10)
b=Num(20)

print(a + b)
```

Substraction

```python
class Num:
  def __init__(self, val):
    self.val=val


  def __sub__(self , obj):
    return self.val - obj.val  ### -> self.val -> a vale object ko and obj.val->b
##object create
a=Num(10)
b=Num(20)

print(a - b)
```

Multiplication

```python
class Num:
  def __init__(self, val):
    self.val=val


  def __mul__(self , obj):
    return self.val *  obj.val
##object create
a=Num(10)
b=Num(20)

print(a * b)
```

Equals than operator

```python
class Num:
  def __init__(self, val):
    self.val=val

## comparision operator
  def __eq__(self , obj):
    return self.val ==  obj.val
##object create
a=Num(2)
b=Num(2)

print(a == b)
```

```python
class Num:
  def __init__(self, val):
    self.val=val
  ##addition
  def __add__(self,obj):
    return self.val+obj.val

  ##substraction
  def __sub__(self,obj):
    return self.val - obj.val

  ## multiplication

  def __mul__(self , obj):
    return self.val * obj.val

  ## division

  def __truediv__(self , obj):
    return self.val / obj.val

  ## modulus
  def __mod__(self,obj):
    return self.val % obj.val


## power
  def __pow__(self , obj):
    return self.val **  obj.val


    ##comparision operator
  def __eq__(self , obj):
    return self.val ==  obj.val
  def __lt__(self , obj):
    return self.val <  obj.val
  def __gt__(self , obj):
    return self.val >  obj.val

##object create
a=Num(25)
b=Num(5)

print("Addition : ",a + b)
print("Substraction : ",a - b)
print("Divison : ",a / b)
print("Modulus : ",a % b)
print("Power : ",a ** b)
print("Multiplication : ",a * b)


print("Less than :",a<b)
print("Greater than :",a>b)
print("Eqaul to :",a==b)
```
