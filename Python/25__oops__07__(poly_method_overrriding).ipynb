```python
class Animal:
  def sound(self):
    print("Animal Sound")
class Dog(Animal):
 def sound(self):
  print("Bow Bow ")
a=Animal()
d=Dog()
a.sound()
d.sound()
```

```python
class Person:
  def display(self):
    print("Person method")
class Student(Person):
  def display(self):
    print("Student Method")
p=Person()
s=Student()
p.display()
s.display()
```

```python
class Shape:
  def area(self):
    print("THis is shape")
class Circle(Shape):
  def area(self):
    print("Area of circle is pi*r*r")
class Rectangle(Shape):
  def area(self):
    print("Area of rectangle l*b")
r=Rectangle()
c=Circle()
r.area()
c.area()
```
