Duck typing -> It does not depend on type of object rather it depends on what object can do

```python
class Book:
  def __init__(self,pageNo,price):
    self.pageNo=pageNo
    self.price=price

  def __len__(self):
   return self.pageNo


book=Book(100,200)
print(len(book))
```

Duck typing -> methods deal

```python
class UPI:
  def pay(self):
    print("This is UPI payment")
class Card:
  def pay(self):
    print("This is card payment")
class Bank:
  def pay(self):
    print("This is Bank payment")



def make_payment(payment):
  payment.pay()

make_payment(UPI())
make_payment(Card())
make_payment(Bank())
```
