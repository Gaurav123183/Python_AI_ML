```python
class calculator:
  def add(self,a,b):
    return a+b
c=calculator()
print("addition before : ",c.add(10,20))


def new_addition(self,a,b):
  return a+b+10


calculator.add=new_addition


print("Addition after: ",c.add(10,20))


```

```python
class calculator:
  def add(self,a,b):
    return a+b
c=calculator()
print("addition before : ",c.add(10,20))


def new_addition(self,a,b):
  return a-b


calculator.add=new_addition

print(c.add(20,10))







```

```python
class calculator:
  def add(self,a,b):
    return a+b
c=calculator()
print("addition before : ",c.add(10,20))


def new_addition(self,a,b):
  return a*b


calculator.add=new_addition

print(c.add(20,10))
```

```python
class calculator:
  def num(self,a):
    return a
c=calculator()
print(c.num(10))


def new_num(self,a):
  return a+100


calculator.num=new_num

print(c.num(10))


```
