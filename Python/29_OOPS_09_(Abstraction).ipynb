```python
from abc import ABC, abstractmethod

class Animal(ABC):
  @abstractmethod
  def sound(self):
    pass

class Dog(Animal):
  def sound(self):
    print("This is Dog")

class Cat(Animal):
  def sound(self):
    print("This is Cat")

class Tiger(Animal):
  def sound(self):
    print("This is Tiger")

class Lion(Animal):
  def sound(self):
    print("This is Lion")

dog=Dog()
dog.sound()
cat=Cat()
cat.sound()
tiger=Tiger()
tiger.sound()
lion=Lion()
lion.sound()
```

```python
from abc import ABC, abstractmethod
class Payment(ABC):
  @abstractmethod
  def pay(self, amount):
    pass
class UPI(Payment):
  def pay(self,amount):
    print("this is upi payment ,paid amount is ",amount)
class Creditcard(Payment):
  def pay(self,amount):
    print("this is creditcard payment ,paid amount is ",amount)
class Netbanking(Payment):
  def pay(self,amount):
    print("this is netbanking payment ,paid amount is ",amount)

upi=UPI()
upi.pay(100)



c1=Creditcard()
c1.pay(1000)


n1=Netbanking()
n1.pay(1500)
```

Aggregation -> Has-A
They can able to access or use another class
But both objects can exist independent


```python
class Student:
  def __init__(self,name):
    self.name=name

class School:
  def __init__(self,Student):
    self.Student=Student

  def show(self):
    print("School has Student ",self.Student.name)

s1=Student("Gaurav")

s2=School(s1)
s2.show()
```
